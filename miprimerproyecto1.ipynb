Ejercicio Encontrar la NF de cada materia, haciendo uso de funciones. Respuesta esperada: PrograI Nota Final 8.5 PrograII Nota Final 7.8

```python
def calcular_nf(persona):
    l1=persona["materias"][0]["notas"]["c1"]["lab1"]
    l2=persona["materias"][0]["notas"]["c1"]["lab2"]
    p3=persona["materias"][0]["notas"]["c1"]["parcial"]
    n_c1=(l1*0.30+l2*0.30+p3*0.40)

    lab1=persona["materias"][0]["notas"]["c2"]["lab1"]
    lab2=persona["materias"][0]["notas"]["c2"]["lab2"]
    par3=persona["materias"][0]["notas"]["c2"]["parcial"]
    n_c2=(lab1*0.30+lab2*0.30+par3*0.40)

    la1=persona["materias"][0]["notas"]["c3"]["lab1"]
    la2=persona["materias"][0]["notas"]["c3"]["lab2"]
    pa3=persona["materias"][0]["notas"]["c3"]["parcial"]
    n_c3=(la1*0.30+la2*0.30+pa3*0.40)
    nota_final = (n_c1+ n_c2 + n_c3)/3

    return nota_final


resultado = calcular_nf(persona)
print("su nota final es:", resultado)
```
