```python
# SUELDO Y DEDUCCIONES

sueldo = float(input("Ingrese el sueldo: $"))

# ISSS = 3%
isss = sueldo * 0.03

# AFP = 6.25%
afp = sueldo * 0.0625


# ISR
if sueldo <= 472:
    isr = 0

elif sueldo <= 895.24:
    isr = (sueldo - 472) * 0.10 + 17.67

elif sueldo <= 2038.10:
    isr = (sueldo - 895.24) * 0.20 + 60

else:
    isr = (sueldo - 2038.10) * 0.30 + 288.61


# Total de deducciones
total = isss + afp + isr

# Sueldo que queda
neto = sueldo - total


# Resultados
print("\n===== RESULTADOS =====")
print("Sueldo:", sueldo)
print("ISSS:", isss)
print("AFP:", afp)
print("ISR:", isr)
print("Total deducciones:", total)
print("Sueldo neto:", neto)
```

```python
print("SISTEMA DE TIEMPO Y DEMANDA UNIDAD DE SALUD")

# Categorias de pacientes
categoria = ["Embarazada", "Paciente tercera edad", "Paciente pediatrico", "Persona adulta/adolescente"]

# Cada sintoma es una tupla: (sintoma, tiempo estimado en minutos, prioridad)
embarazada = [
    ("Riesgo de hipertension (preeclampsia)", 30, "Alta"),
    ("Nauseas y vomitos (mareos matutinos)", 15, "Media"),
    ("Dolor lumbar y molestias pelvicas", 15, "Media"),
]

paciente_tercera_edad = [
    ("Hipertension arterial y enfermedades cardiacas", 30, "Alta"),
    ("Diabetes tipo 2", 25, "Alta"),
    ("Artritis y artrosis (dolores articulares)", 20, "Media"),
]

paciente_pediatrico = [
    ("Fiebre alta repentina", 25, "Alta"),
    ("Gastroenteritis (diarrea y vomitos)", 20, "Media"),
    ("Resfriado comun, gripe y congestion nasal", 15, "Baja"),
]

persona_adulta_adolescente = [
    ("Hipertension o diabetes de inicio temprano", 30, "Alta"),
    ("Estres, ansiedad y depresion", 25, "Media"),
    ("Dolores de cabeza y migranas", 15, "Media"),
]

opcion = [embarazada, paciente_tercera_edad, paciente_pediatrico, persona_adulta_adolescente]

# Lista donde se van agregando los pacientes registrados
listado_prioridad = []

# Orden numerico de cada prioridad, para poder ordenar la lista despues
orden_prioridad = {"Alta": 1, "Media": 2, "Baja": 3}

seguir = "s"
while seguir == "s":
    nombre = input("\nIngrese su nombre: ")
    edad = int(input("Ingrese su edad: "))
    peso = float(input("Ingrese su peso en kilogramos: "))
    altura = float(input("Ingrese su altura en metros: "))
    sexo = input("Ingrese su sexo (M para masculino, F para femenino): ")

    # IMC del paciente
    imc = peso / (altura ** 2)
    if imc < 18.5:
        clasificacion_imc = "Bajo peso"
    elif imc < 25:
        clasificacion_imc = "Peso normal"
    elif imc < 30:
        clasificacion_imc = "Sobrepeso"
    else:
        clasificacion_imc = "Obesidad"

    print("Seleccione una de las siguientes opciones")
    for i, cat in enumerate(categoria, start=1):
        print(f"{i}. {cat}")
    eleccion = int(input("Seleccione una opcion: "))

    if eleccion == 1:
        print(">>>>>Paciente embarazada<<<<<")
        for i, (sintoma, tiempo, prioridad) in enumerate(embarazada, start=1):
            print(f"{i}. {sintoma}")
        seleccion = int(input("Seleccione una opcion: "))
        sintoma_elegido, tiempo_estimado, prioridad = embarazada[seleccion - 1]
        nombre_categoria = "Embarazada"

    elif eleccion == 2:
        print(">>>>>Paciente de la tercera edad<<<<<")
        for i, (sintoma, tiempo, prioridad) in enumerate(paciente_tercera_edad, start=1):
            print(f"{i}. {sintoma}")
        seleccion = int(input("Seleccione una opcion: "))
        sintoma_elegido, tiempo_estimado, prioridad = paciente_tercera_edad[seleccion - 1]
        nombre_categoria = "Paciente de la tercera edad"

    elif eleccion == 3:
        print(">>>>>Paciente pediatrico<<<<<")
        for i, (sintoma, tiempo, prioridad) in enumerate(paciente_pediatrico, start=1):
            print(f"{i}. {sintoma}")
        seleccion = int(input("Seleccione una opcion: "))
        sintoma_elegido, tiempo_estimado, prioridad = paciente_pediatrico[seleccion - 1]
        nombre_categoria = "Paciente pediatrico"

    elif eleccion == 4:
        print(">>>>>Persona adulta/adolescente<<<<<")
        for i, (sintoma, tiempo, prioridad) in enumerate(persona_adulta_adolescente, start=1):
            print(f"{i}. {sintoma}")
        seleccion = int(input("Seleccione una opcion: "))
        sintoma_elegido, tiempo_estimado, prioridad = persona_adulta_adolescente[seleccion - 1]
        nombre_categoria = "Persona adulta/adolescente"

    else:
        print("Opcion invalida")
        seguir = input("\n¿Desea agregar otro paciente? (s/n): ").lower()
        continue

    # Se agrega el paciente a la lista, como un diccionario con sus datos
    listado_prioridad.append({
        "nombre": nombre,
        "categoria": nombre_categoria,
        "sintoma": sintoma_elegido,
        "tiempo": tiempo_estimado,
        "prioridad": prioridad
    })

    seguir = input("\n¿Desea agregar otro paciente? (s/n): ").lower()

# Se ordena la lista de pacientes segun su prioridad (Alta primero, luego Media, luego Baja)
listado_prioridad.sort(key=lambda paciente: orden_prioridad[paciente["prioridad"]])

# Se muestra el listado final en pantalla
print("\n===== LISTADO DE PACIENTES POR PRIORIDAD =====")
for i, paciente in enumerate(listado_prioridad, start=1):
    print(f"{i}. {paciente['nombre']}, {paciente['categoria']}, prioridad: {paciente['prioridad']}")
```
